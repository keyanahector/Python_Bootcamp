#### 1. Write a conditional that prints different messages if a bank account balance is below 3,000, between 3,000 and 10,000, or over 10,000.

```python
def bank_account_balance():
  bank_account_balance = int(input("what is your bank account balance?"))
  if bank_account_balance < 3000:
    print("your balance is low")
  elif bank_account_balance <= 10000:
    print("your balance is medium")
  elif bank_account_balance > 10000:
    print("your balance is high")
  else:
    print("your balance is overdrafted")
```

#### 1. Using `w`, create a file called fruits.txt and write the string "tangerines" to it.

```python
def fruit_name():
  fruit = input("Enter a fruit name:")
  with open("fruits.txt", "w") as file:
    file.write(fruit)
    return("File has been created!")
```
